### T-Period Parent-Child Interaction Model with Human Capital Investment

Solves a $T$-period dynamic model where a family, consisting of a parent and a childakes decisions over parental consumption, child's study time, education expenditure, parental labor supply, and child care time to maximize a weighted sum of their lifetime utilities. The utility function incorporates parental consumption, leisure for both parent and child, and the child's human capital, reflecting preferences over these elements and investment in the child's future human capital.

#### Model Description

At each period $t = 1, \dots, T_L-1$, the family chooses:

- **Parental consumption** $c_{p,t} > 0$
- **Child's study time** $\tau_{c,t} \in [0, 1]$
- **Education expenditure** $e_{p,t} > 0$
- **Parental labor supply** $h_{p,t} \in [0, 1]$
- **Child care time** $\tau_{p,t} \in [0, 1]$

Define the following leisure variables:

- **Parental leisure:** $$l_{p,t} = 1 - h_{p,t} - \tau_{p,t}$$
- **Child’s leisure:** $$l_{c,t} = 1 - \tau_{p,t} - \tau_{c,t}$$

With this shocks in the model:
- **AR(1) shock** $$z_{t+1} = \rho z_t + \eta_{t+1}, \quad \eta_{t+1} \sim \mathcal{N}(0, \sigma^2_z)$$
- **Preference shock** $$\epsilon_0 \sim \mathcal{N}(0, \sigma^2_{\epsilon})$$

to maximize the value function:

- **Utility of the parent:**
  $$
  U_{p,t} = \phi_1 \log c_{p,t} + \phi_2 \log l_{p,t} + \phi_3 \log HC_t
  $$
- **Utility of the child:**
  $$
  U_{c,t} = \lambda_1 \log l_{c,t} + \lambda_2 \log HC_t
  $$

- **Utility of the family:**

  $$
  U_{f,t} =
      \phi_1 \log c_{p,t}
    + \phi_2 \log l_{p,t}
    + \tilde{\alpha}_1 \log l_{c,t}
    + \tilde{\alpha}_2 \log HC_t,
  $$
    -where
    $$
    \begin{aligned}
      \tilde{\alpha}_1 &= (1 - \tilde{\mu}_t) \cdot \lambda_1, \\
      \tilde{\alpha}_2 &= \tilde{\mu}_t \cdot \phi_3 + (1 - \tilde{\mu}_t) \cdot \lambda_2.
    \end{aligned}
    $$

- **Bargaining weight:**
$$
\tilde{\mu}_t =
\begin{cases}
= 1           & \text{if } t \leq 7 \\ 
= 1 - \mu_1 (t-7) & \text{if } 7 < t \leq 17
\end{cases}
$$


- **Value function:**


  $$
  V_t(a_t, k_t, HC_t) =
  \begin{cases}
    \displaystyle\max_{c_{p,t},\,e_{p,t},\,h_{p,t},\,\tau_{p,t}}
      \left\{ U_{f,t} + \beta \mathbb{E}_{z_t+1} \left[ V_{t+1}(a_{t+1}, k_{t+1}, HC_{t+1}) \right]\right\}
      & \text{if } t \leq 7 \\[4ex]
    \displaystyle\max_{c_{p,t},\,\tau_{c,t},\,e_{p,t},\,h_{p,t},\,\tau_{p,t}}
      \left\{ U_{f,t} + \beta \mathbb{E}_{z_t+1} \left[ V_{t+1}(a_{t+1}, k_{t+1}, HC_{t+1}) \right]\right\}
      & \text{if } 7 < t < 17 \\[4ex]
    \displaystyle\max_{c_{p,t},\,\tau_{c,t},\,e_{p,t},\,h_{p,t},\,\tau_{p,t}}
      \left\{ U_{f,t} + \beta\, \mathbb{E}_{z_0, \epsilon_0}\left[ V_{T_L-1}^{\text{CD}}(a_{t+1}, HC_{t+1}) \right] \right\}
      & \text{if } t = 17
  \end{cases}
  $$


subject to the following constraints:



- **Budget constraint**:
  $$
  a_{t+1} = (1 + r)\, a_t + (1 - \tau_{\text{tax}})\, w_t\, h_{p,t} + b_t - c_{p,t} - e_{p,t} 
  $$
  $$
  a_{t+1} \geq 0
  $$

- **Time constraints**:
  - Parent's time:
    $$
    h_{p,t} + \tau_{p,t} + l_{p,t} = 1.0
    $$
  - Child's time:
    $$
    \tau_{c,t} + \tau_{p,t} + l_{c,t} =  1.0
    $$

- **Parent's human capital accumulation**:
  $$
  k_{t+1} = k_t + h_{p,t}
  $$
- **Child's human capital technology**:
$$
\log HC_{t+1} =
\begin{cases}
\log R_t + \sigma_{1t} \log \tau_{p,t} + \sigma_{2t} \log e_{p,t} + \sigma_{3t} \log HC_t & \text{if } t \leq 7 \\[2ex]
\log R_t + \sigma_{1t} \log \tau_{p,t} + \sigma_{2t} \log e_{p,t} + \sigma_{3t} \log HC_t + \sigma_{4t} \log \tau_{c,t} & \text{if } 7< t \leq 17 
\end{cases}
$$
- **Wage function**:
  $$
  w_t = w_0 \cdot (1 + \alpha k_t)
  $$

---

  **Terminal Value at Period $T_L$ and College Decision Half period**

  The parent utility in period ($T = T_L$) For an agent with belief $m$ is defined as: 
  $$
  V_{T_L}^P = \underbrace{\psi_{\text{term}} \cdot \log HC_{T_L}}_{\text{HC Prestige}} + \kappa_{\text{term}} \cdot \log a_{term} + \underbrace{\omega \cdot V_{T_L}^C}_{\text{Altruism}}
  $$

The child is discrete choice is $d \in \{E,W\}$:

  $$
  V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L}) = \max \left\{ V_0^E(tr, HC_{T_L}, \epsilon_0, z_{T_L}), V_0^W(tr, HC_{T_L}, z_{T_L}) \right\}
  $$



The family’s ex-ante value function 


$$
V_{T_L}^{\mathrm{CD}}(a_{T_L}, HC_{T_L}, \epsilon_0) =
\max_{0 \leq tr \leq a_{T_L}} \left\{
\mathbb{E}_{z_{T_L}} \left[
(1 - \theta) V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L})
+ \theta V_{T_L}^P(tr, HC_{T_L}, \epsilon_0)
\right]
\right\}
$$



- in period $T_L-1$ we had:
    $$
    V_{T_L-1}^{\mathrm{CD}}(a_{T_L}, HC_{T_L}) =
    \max_{0 \leq tr\leq a_{T_L}} \left\{
        \mathbb{E}_{\epsilon_0,\, z_{T_L}} \left[
            (1 - \theta) V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L})
            + \theta V_{T_L}^P(tr, HC_{T_L}, \epsilon_0)
        \right]
    \right\}
    $$

The decision will be: 
$$
d^*(a_{T_L}, HC_{T_L}, \epsilon_0) =
     \arg\max_{\substack{0 \leq tr \leq a_{T_L}}} \left\{ 
            \mathbb{E}_{z_{T_L}} \left[
                (1-\theta)\, V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L})
                + \theta\, V_{T_L}^P(tr, a_{T_L}, HC_{T_L}, \epsilon_0)            \right]
        \right\}
$$

subject to the following constraints:



- **Budget constraint**:
  $$
  a_{term} = a_{T_L} - tr
  $$

### Utility of Child after Sepration

The transfer in the terminal period become the child's intial asset. from $T_L$ to $T_D $ (age 18 to 68) we have:


**College ($s_t = E$)**

$$
V^E_t(a_t, HC_t, \epsilon_0, z_t) =
\begin{cases}
  \displaystyle
  \max_{c_t}
  \left\{ \frac{c_t^{1-\rho}}{1-\rho}
      - \kappa_X
      + \epsilon_0
      + \beta\, \mathbb{E}_{z_{t+1}}\left[ V^E_{t+1}(a_{t+1}, HC_{t+1}, z_{t+1}) \right]
  \right\}
    & \text{if } t = 1 \\[5ex]

  \displaystyle
  \max_{c_t}
  \left\{ \frac{c_t^{1-\rho}}{1-\rho}
      - \kappa_X
      + \beta\, \mathbb{E}_{z_{t+1}}\left[ V^E_{t+1}(a_{t+1}, HC_{t+1}, z_{t+1}) \right]
  \right\}
    & \text{if } 2 \leq t < 4 \\[5ex]

  \displaystyle
  \max_{c_t}
  \left\{ \frac{c_t^{1-\rho}}{1-\rho}
      - \kappa_X
      + \beta\, \mathbb{E}_{z_{t+1}}\left[ V^W_{t+1}(a_{t+1}, HC_{t+1}, z_{t+1}) \right]
  \right\}
    & \text{if } t = 4
\end{cases}
$$

$$
a_{t+1} = (1 + r) a_t - c_t - c_{\text{college}} + b_t
$$

$$
\kappa_X = \kappa \frac{1}{(HC_t+1)^2}
$$

$$
HC_{t+1} =
\begin{cases}
HC_t + h_t^E & \text{for } t \leq t_{\text{college}} \\
HC_t + h_t   & \text{for } t > t_{\text{college}}
\end{cases}
$$



**Labor Market ($s_t = W$)**

$$
V^W_t(a_t, HC_t, z_t) = \max_{c_t, h_t} \left\{ \frac{c_t^{1-\rho}}{1-\rho} -\varphi \frac{h_t^{1+\eta}}{1+\eta} + \beta\, \mathbb{E}_{z_{t+1}} \left[ V^W_{t+1}(a_{t+1}, HC_{t+1}, z_{t+1}) \right] \right\}
$$

$$
a_{t+1} = (1 + r)a_t + (1 - \tau_{\text{tax}})\, w_t\, h_t - c_t + b_t
$$



$$
HC_{t+1} = HC_t + h_t
$$

**Wage Process (with Shocks)**
$$
w_t = w^{\text{base}}_t \cdot (1 + \alpha \cdot HC_t) \cdot z_t 
$$

### Pseudo-Algorithm

The following outlines the steps for solving the age-specific dynamic family model using value function iteration and simulating agent behavior over time.

#### **Model Solution (Backward Induction)**

**Input:**  
Parameter set, grids for assets ($a$), parental human capital ($k$), child human capital ($HC$), terminal utility function $V_T$

**For** $t = T$ **down to** $1$:

1. **If** $t = T_{\text{terminal}}$ (Separation/Terminal Period):
    - **For each grid point** $(a_T, k_T, HC_T)$:
        - Define bounds for decision variables (consumption, study time, education expenditure, labor, care time)
        - Set nonlinear constraints (minimum leisure, feasible time allocation, asset/budget)
        - **Solve:**
            $$
            \max_{\text{controls}} \Big\{ U_T(\cdot) + \beta_T V_{T+1}(a_{T+1}, k_{T+1}, HC_{T+1}) \Big\}
            $$
            where
            $$
            V_{terminal} = \underbrace{\psi_{\text{terminal}} \cdot \log HC_{T+1}}_{\text{HC Prestige}} + \kappa_{\text{termianl}} \cdot \log a_{T+1}

            $$

        - Add Altruism:
            $$
            \omega\,\mathbb{E}\left[V^{\text{child}}(tr, HC_{t+1})\right] =\omega \cdot \mathbb{E}_{\text{wage, preference shocks}} \big[ V^{\text{child}}(tr, HC_{t+1}) \big]
            $$
            where the expectation is over both:
                - Markov wage shocks (future persistent wage states) and i.i.d. normal preference (“taste”) shocks affecting the college vs. work decision
        - **Store:**
            - Policy functions for all controls at $(a_T, k_T, HC_T)$
            - Value function $V_T(a_T, k_T, HC_T)$

2. **Else:** (For $t < T_{\text{terminal}}$)
    - **For each grid point** $(a_t, k_t, HC_t)$:
        - Define bounds for decision variables (consumption, study time, education expenditure, labor, care time)
        - Set nonlinear constraints (minimum leisure, feasible time allocation, asset/budget)
        - **If** ($t > 7$, **Post-age 7**):
            - Use utility and human capital update equations relevant for $t > 7$
            - **Solve:**
                $$
                \max_{\text{controls}} \Big\{ U_T(\cdot) + \beta_t V_{t+1}(a_{t+1}, k_{t+1}, HC_{t+1}) \Big\}
                $$
                where
                $$
                \tilde{\mu}_t = 1 - \mu_1 (t-7),\quad
                \log HC_{t+1} = \log R_t + \sigma_{1t} \log \tau_{p,t} + \sigma_{2t} \log e_{p,t} + \sigma_{3t} \log HC_t + \sigma_{4t} \log \tau_{c,t}
                $$
        - **Else** $t \leq 7$ (**Pre-age 7**):
            - Use utility and human capital update equations relevant for $t \leq 7$
            - **Solve:**
                $$
                \max_{\text{controls}} \Big\{ U_T(\cdot) + \beta_t V_{t+1}(a_{t+1}, k_{t+1}, HC_{t+1}) \Big\}
                $$
                where
                $$
                \tilde{\mu}_t = 1,\quad
                \log HC_{t+1} = \log R_t + \sigma_{1t} \log \tau_{p,t} + \sigma_{2t} \log e_{p,t} + \sigma_{3t} \log HC_t
                $$
        - **Store:**
            - Policy functions for all controls at $(a_t, k_t, HC_t)$
            - Value function $V_t(a_t, k_t, HC_t)$

**Output:**  
Grids of value and policy functions for all periods and state combinations

### Baseline and Counterfactuals: Human Capital Coefficients and Terminal Value Effects

**Reading these policy functions (checked 9 September 2026).** The next figure plots
one baseline policy slice over assets, not counterfactual differences or population
averages. It holds BothCollege at 0, HC at grid node 15 (408.3), and productivity
at its **lowest** node (0.632). These states are now exposed as keyword arguments
and printed in the figure title. Assets are in thousands of dollars; the axis
extends to $1 million and the nonlinear grid packs points near zero.

Higher wealth supports more consumption and less work. At age 6 it also raises
parental time, which competes with the child's time in the shared child time budget.
For an interior child-time choice, letting $B=HC'\,\partial V_{t+1}/\partial HC'$, its
first-order condition is
$$
\frac{(1-\mu_t)\lambda_1}{1-\tau_{p,t}-i_t-s_t}
=\frac{\beta_t B\sigma_{4,t}}{i_t}.
$$
When $B$ changes little with wealth, more parental time leaves less room for child
study and leisure. Thus declining child investment alongside rising parental time
is consistent with this model. Here $s_t=0$ for the fitted legacy total-input model;
future own-study runs use a separate fixed school schedule.

Terminal policies depend instead on the transfer/college/adult-life value surface.
The falling part of terminal education spending reflects changes in the relative
marginal values of HC and assets, rather than an imposed monotonicity rule. For this
slice, $B$ falls from about 4.42 to 1.01 between the lowest and penultimate asset nodes.
At the final node, next-period assets equal the **100-model-unit ceiling**; the jump
in consumption there is a binding grid/budget restriction, not a smooth wealth effect.

Small terminal spending zigzags should not be interpreted economically. Re-optimizing
the same objective at node 9 with `xtol_rel=1e-8` changes spending from about $98 to
$149, for an objective gain of only 0.000376. The displayed baseline keeps the existing
solver and fitted calibration; a converged return code alone does not ensure equally
accurate policy levels. No smoothing or solver retuning has been used to conceal this.

The original stored figures predated the present calibration and HC units (their
terminal mean HC was 1.44). The two figures in this section have been refreshed from
the current fitted baseline, using child/parent grids of 30, `Nk=2` for parents,
`Nt=5` for the child, `simN=5000`, seed 1234 and terminal spline `s=10`.

```julia
using DelimitedFiles

# -----------------------------------------------------------------------------
# Data means by child age -- TWO samples, for the overlays below
# -----------------------------------------------------------------------------
# Generated by tools/make_smm_targets.py. Julia has no Stata reader in this project, so
# Python touches the .dta and Julia reads small tracked CSVs -- the same rule the
# estimation targets already follow.
#
# WHY TWO. The model assumes parents are aged 26 at the child's birth, which is exactly
# what the cohort file restricts to (cohort_dad2530, father 25-30 at birth). The other
# file has no parental-age restriction. That restriction is NOT cosmetic:
#
#     moment                    all families    cohort      diff
#     assets (ex-housing)          191,200      137,672    -28.0%
#     monetary investment            3,945        3,243    -17.8%
#     consumption                   31,577       29,162     -7.7%
#     parent time with child         43.39        45.00     +3.7%
#     work hours                     84.55        84.64     +0.1%
#
# So the money moments move a lot under the restriction and the time moments barely at
# all -- and the SMM targets are currently built from the UNRESTRICTED sample. Plotting
# both makes the size of that gap visible rather than leaving it an assumption.
#
# GUARDED: the .dta files are gitignored, so a checkout that has not run the generator
# simply plots whatever it has, or the model alone.
function load_byage(fname)
    path = joinpath(ROOT, "Input", fname)
    try
        raw, hdr = readdlm(path, ',', header = true, comments = true)
        cols = vec(String.(hdr))
        col(n) = (j = findfirst(==(n), cols);
                  j === nothing ? fill(NaN, size(raw, 1)) :
                  [v isa Number ? Float64(v) : NaN for v in raw[:, j]])
        (age = Int.(col("child_age")), c_p = col("c_p"), e_p = col("e_p"), a_p = col("a_p"),
         h_p = col("h_p"), t_p = col("t_p"), i_c = col("i_c"),
         i_total = col("i_total"), school_c = col("school_c"),
         x_gach = col("x_gach"), x_lw = col("x_lw"))
    catch err
        @warn "Missing by-age file; that series will be omitted. Regenerate with tools/make_smm_targets.py" file=path
        nothing
    end
end

BYAGE     = load_byage("smm_moments_by_age.csv")
BYAGE_COH = load_byage("smm_moments_by_age_cohort.csv")

DATA_SETS = filter(x -> x[1] !== nothing,
    [(BYAGE,     "Data, all families", :dash,    :black),
     (BYAGE_COH, "Data, cohort 25-30", :dashdot, :grey35)])

"""
Overlay both samples for one field. `scale` converts model units to the panel's units.
`targeted = false` marks a series the SMM did NOT match, so a gap there is information
rather than a miss.
"""
function overlay!(pl, field::Symbol; targeted::Bool = true, scale::Real = 1,
                  sets = DATA_SETS)
    for (ds, name, style, colour) in sets
        y = getfield(ds, field) .* scale
        ok = findall(isfinite, y)
        isempty(ok) && continue
        plot!(pl, ds.age[ok], y[ok];
              label = name * (targeted ? "" : " (not targeted)"),
              lw = 1.0, ls = style, color = colour, alpha = 0.8)
    end
    return pl
end

# Every data value for one field, for the y-limits.
alldata(field::Symbol; scale::Real = 1, sets = DATA_SETS) =
    isempty(sets) ? Float64[] :
    vcat((getfield(ds, field) .* scale for (ds, _, _, _) in sets)...)

# The all-families sample on its own. Selected BY NAME, not by index: DATA_SETS drops any
# sample whose CSV is missing, so DATA_SETS[1] is not reliably the unrestricted one.
ALL_FAMILIES_ONLY = filter(x -> occursin("all families", x[2]), DATA_SETS)

# A y-limit that spans the data too. Without this a data point above the model's maximum is
# clipped and silently invisible -- the failure that hid the baseline lines in the
# counterfactual panels earlier.
topof(mv, dv) = 1.1 * maximum(filter(isfinite, vcat(collect(Float64, mv), collect(Float64, dv))))

# --- Extraction Function ---
function extract_all(model)
    return (
        c      = mean(model.sim_c, dims=1)[:],
        a      = mean(model.sim_a, dims=1)[:],      # length T+1
        h      = mean(model.sim_h, dims=1)[:],
        k      = mean(model.sim_k, dims=1)[:],      # length T+1
        t      = mean(model.sim_t, dims=1)[:],
        e      = mean(model.sim_e, dims=1)[:],
        i      = mean(model.sim_i, dims=1)[:],
        hc     = mean(model.sim_hc, dims=1)[:],     # length T+1, in LEVELS
        # The cross-agent mean of LOG hc, which is a different object from log of the
        # mean. HC is a LEVEL in this model (HC_technology returns exp(...), utility
        # takes log(HC)), and the data's x_gach is a mean of
        # LOGS -- so comparing log(mean hc) against mean(log score) would carry a
        # Jensen gap of about Var(log hc)/2 that moves with age. Match the data.
        log_hc = mean(log.(max.(model.sim_hc, 1e-12)), dims=1)[:],
        income = mean(model.sim_income, dims=1)[:],
        wage   = mean(model.sim_wage, dims=1)[:]
    )
end


function plot_parent_child_before_separation(
    model_baseline;
    plot_title::AbstractString = "Parent-Child Variables Before Separation",
    rescale::Real = 10,
    tick_step_flow::Int = 2,
    tick_step_state::Int = 2,
    fig_size::Tuple{Int,Int} = (1500, 1800),
    dpi::Int = 100,
    save::Bool = false,
    save_dir::Union{Nothing,AbstractString} = nothing,
    include_title_in_file::Bool = false
)
    # ---- data ----
    base = extract_all(model_baseline)
    T = model_baseline.T
    t = 1:T
    t_state = 1:(T+1)

    # scaled variables
    c_plot      = base.c .* rescale
    a_plot      = base.a[1:T+1] .* rescale
    income_plot = base.income .* rescale
    e_plot      = base.e .* rescale
    hc_plot     = base.hc[1:T+1]
    k_plot      = base.k[1:T+1]

    # y-lims
    ymax_c      = topof(c_plot, alldata(:c_p; scale = rescale))
    ymax_a      = maximum(a_plot)      * 1.1
    ymax_income = maximum(income_plot) * 1.1
    ymax_wage   = maximum(base.wage)   * 1.1
    ymax_e      = topof(e_plot, alldata(:e_p; scale = rescale))
    ymax_hc     = maximum(hc_plot)     * 1.1
    ymax_k      = maximum(k_plot)      * 1.1
    ymax_h      = topof(base.h, alldata(:h_p))
    ymax_t      = topof(base.t, alldata(:t_p))
    # The fitted baseline still uses the legacy total input; a newly estimated
    # model carries positive fixed school time and chooses own study only.
    # The child chooses OWN STUDY; school is a fixed time cost it cannot trade away, so
    # the data counterpart is `i_c` (study_hrs/112), not `i_total` (school + study).
    # school_time now defaults to the real schedule, so own_study is true unless a caller
    # deliberately passes zeros(T) to recover the pre-2026-09-09 behaviour.
    own_study = any(>(0), model_baseline.school_time)
    child_time_field = own_study ? :i_c : :i_total
    # ONE sample on this panel, by instruction: own study is measured on the CDS time
    # diary, whose per-age cell counts are the smallest of any moment (36-47 in the
    # unrestricted sample), and the cohort restriction cuts them further. Two noisy dashed
    # lines around a flat series read as disagreement in the data rather than as sampling
    # error. The other panels keep both samples.
    child_time_sets = ALL_FAMILIES_ONLY
    ymax_i = topof(base.i, alldata(child_time_field; sets = child_time_sets))

    thousand_formatter2(y) = string(round(y; digits=2))
    style_grid!(p) = plot!(p; grid=:on, gridstyle=:dash,
        foreground_color_grid=RGBA(0.5,0.5,0.5,0.85), gridlinewidth=1, minorgrid=false)

    xticks_flow  = 1:tick_step_flow:T
    xticks_state = 1:tick_step_state:(T+1)

    # ---- individual panels ----
    p1  = plot(t, c_plot, label="Consumption", lw=2,
               xlabel="Child age", ylabel="Consumption (×10³)",
               yformatter=thousand_formatter2, ylim=(0, ymax_c), xticks=xticks_flow);  style_grid!(p1)
    overlay!(p1, :c_p; scale = rescale)

    # Assets: data is PSID family net worth EXCLUDING home equity, which is what the model
    # can be compared against -- it has no housing sector, no mortgage and no durable
    # stock, so home equity has nothing to map onto (the same reason consumption uses
    # cons_exhous_real). UNTARGETED: the SMM never matched assets, so a gap here is a
    # statement about the model, not a failed fit.
    p2  = plot(t_state, a_plot, label="Parent Assets", lw=2,
               xlabel="Child age", ylabel="Assets (×10³)",
               yformatter=thousand_formatter2,
               ylim=(min(0, minimum(a_plot)), topof(a_plot, alldata(:a_p; scale = rescale))),
               xticks=xticks_state); style_grid!(p2)
    overlay!(p2, :a_p; targeted = false, scale = rescale)

    p3  = plot(t, base.h, label="Hours Worked", lw=2,
               xlabel="Child age", ylabel="Hours", ylim=(0, ymax_h),
               xticks=xticks_flow); style_grid!(p3)
    overlay!(p3, :h_p)

    # sim_k is the BothCollege INDICATOR, not human capital: a binary household type drawn
    # once from Bernoulli(0.3) and constant in t, so this line is the college share and is
    # flat by construction. It was labelled "Parental HC", which is the same mislabel the
    # struct carried. There is no data counterpart -- it is an input, not an outcome.
    p4  = plot(t_state, k_plot, label="BothCollege share", lw=2,
               xlabel="Child age", ylabel="Parents both college (share)", ylim=(0, ymax_k),
               xticks=xticks_state); style_grid!(p4)

    p5  = plot(t, e_plot, label="Education Exp.", lw=2,
               xlabel="Child age", ylabel="Education Exp. (×10³)",
               yformatter=thousand_formatter2, ylim=(0, ymax_e),
               xticks=xticks_flow); style_grid!(p5)
    overlay!(p5, :e_p; scale = rescale)

    # HC now uses the data's W-score units. Compare mean(log HC) directly;
    # shifting each data line to the model would hide the HC level moments' fit.
    log_hc = base.log_hc[1:T+1]
    p6 = plot(t_state, log_hc, label="Child's HC (mean log)", lw=2,
              xlabel="Child age", ylabel="Mean log human capital",
              xticks=xticks_state); style_grid!(p6)
    for (ds, name, style, colour) in DATA_SETS
        ok = findall(j -> ds.age[j] >= 3 && isfinite(ds.x_gach[j]), eachindex(ds.age))
        isempty(ok) && continue
        plot!(p6, ds.age[ok], ds.x_gach[ok];
              label=name, lw=1.0, ls=style, color=colour, alpha=0.8)
    end

    p7  = plot(t, base.t, label="Childcare Time", lw=2,
               xlabel="Child age", ylabel="Childcare Time", ylim=(0, ymax_t),
               xticks=xticks_flow); style_grid!(p7)
    overlay!(p7, :t_p)

    p8  = plot(t, income_plot, label="After-tax labor income", lw=2,
               xlabel="Child age", ylabel="Labor income (×10³)",
               yformatter=thousand_formatter2, ylim=(0, ymax_income),
               xticks=xticks_flow); style_grid!(p8)

    p9  = plot(t, base.wage, label="Wage", lw=2,
               xlabel="Child age", ylabel="Wage", ylim=(0, ymax_wage),
               xticks=xticks_flow); style_grid!(p9)

    p10 = plot(t, base.i, label=own_study ? "Own study" : "School + study (legacy)", lw=2,
               xlabel="Child age", ylabel="Child's Time Inv.",
               yformatter=thousand_formatter2, ylim=(0, ymax_i),
               xticks=xticks_flow); style_grid!(p10)
    overlay!(p10, child_time_field; sets = child_time_sets)

    # Give every legend its own space above its axes. Both display and saved
    # figures use these panels, so no legend can cover a trajectory.
    for panel in (p1, p2, p3, p8, p9, p7, p5, p4, p10, p6)
        plot!(panel; legend=:outertop, legendfontsize=8,
              legend_columns=1, background_color_legend=:transparent,
              foreground_color_legend=:transparent)
    end

    # ---- combined: display (with title) ----
    fig_display = plot(p1, p2, p3, p8, p9, p7, p5, p4, p10, p6;
        layout=(5, 2), size=fig_size, plot_title=plot_title,
        titlefont=16, margin=5mm, left_margin=12mm, dpi=dpi)

    display(fig_display)

    # ---- combined: save (optionally without title) ----
    fig_save = (!save || include_title_in_file) ? fig_display :
        plot(p1, p2, p3, p8, p9, p7, p5, p4, p10, p6;
             layout=(5, 2), size=fig_size, margin=5mm, left_margin=12mm, dpi=dpi)

    # ---- saving ----
    if save
        # choose/prepare folder
        if save_dir === nothing
            stamp = Dates.format(Dates.now(), "yyyy-mm-dd_HHMMSS")
            save_dir = figdir("Baseline", stamp)
        end
        isdir(save_dir) || mkpath(save_dir)

        # safe filename from title
        safe_title = replace(replace(String(plot_title), r"[^\w\s-]" => ""), " " => "_")
        outpath = joinpath(save_dir, safe_title * ".pdf")

        # avoid overwrite
        if isfile(outpath)
            i = 2
            while isfile(joinpath(save_dir, "$(safe_title)_$(i).pdf")); i += 1; end
            outpath = joinpath(save_dir, "$(safe_title)_$(i).pdf")
        end

        savefig(fig_save, outpath)   # PDF (vector)
        @info "Saved figure as" outpath
    else
        @info "Plot not saved (save=false)"
    end

    return fig_display
end

plot_parent_child_before_separation(model_baseline;
    save=false,
    save_dir=figdir("Baseline"),
    include_title_in_file=false);
```

### Why the last period moves, and the specification for future runs

The refreshed fitted baseline has the following age-16 to age-17 movement:

| Variable | Age 16 | Age 17 |
|---|---:|---:|
| Education expenditure, dollars/year | 2,438 | 170 |
| Child time input, share of 112-hour week (legacy school + study) | 0.4662 | 0.1028 |
| Parental time with child, share | 0.2225 | 0.0228 |
| Parental work, share | 0.4670 | 0.6560 |
| After-tax labor income, dollars/year | 71,772 | 95,329 |

**The continuation changes at separation.** Age-16 investment raises age-17 HC,
which still earns family flow utility. The coefficient on `log(HC)` at age 17 is
`mu[17]*phi_3 + (1-mu[17])*lambda_2`, about **29.65** in this fitted baseline.
Age-17 investment raises age-18 HC, which is valued through the transfer/college/adult
continuation instead. Across the grid states with BothCollege=0, the median marginal
continuation value with respect to log HC falls from **30.62 at age 16 to 2.58 at 17**.
This substantially reduces the incentive to invest at 17. The money elasticity is
also already very small (about 0.00111 at 17). The old saved figure's upward spike
belongs to a different calibration; the switch in continuation can produce either
direction, depending on its marginal values.

**Income follows work choices as well as wages.** `sim_income` is
`tax_lambda * (wage_func(...) * sim_h)^(1-tau)`, averaged over households. It excludes
the fixed transfer `y` and asset interest. Wages rise smoothly here; childcare and
work change sharply, so labor income need not be smooth or hump-shaped. The dip near
age 5 is consistent with anticipatory HC investment before the child's utility and own input
activate at 6; the final rise accompanies the collapse in parental investment time.
This is model behavior, not an observed-income overlay.

**Presentation corrections.** Legends now sit outside each axes, above the lines,
in a five-row/two-column layout. HC is plotted as mean log HC in W-score units, without
shifting the data to the model; the former shift hid level-fit differences. The
child-time overlay selects the legacy total input for this fitted baseline and own
study for models with a positive fixed school schedule.

**Confirmed future specification.** `i_t` is own study only (`study_hrs/112`); fixed
school time is separately deducted in `l_c = 1 - tau_p - i_t - school_time[t]`.
Child leisure utility remains logarithmic, and only own study enters the child-time
term in production. `school_hrs` contains the median within `(Year, Child_Age)`;
the fixed age schedule is the mean of those nonmissing values at each age, divided by
112. It is frozen in the generated target file. `c_time_hrs - study_hrs` is not used,
because the total series includes imputations and has different coverage.

Own-study targets cover ages 6–9 (**4.40 hours/week**) and 10–17 (**5.55 hours/week**).
The early HC moment remains mean log HC at 3–9; the late HC moment now covers **12–17**
(target **6.2589**). `sigma_4_1` is the tenth estimated parameter, using the existing
diagnostic candidate box `[-0.05, 0.15]`; `mu_1` remains fixed. Ten parameters and ten
moments require a fresh identification check after estimation.

The existing `PARENT_DEFAULTS` and notebook baseline retain the fitted total-input
specification. SMM explicitly passes the new school schedule into every objective,
fit report and sensitivity/profile solve. Regenerate targets and start a **new run**;
old target files and nine-parameter checkpoints must not be reused with new semantics.
See [the SMM instructions](smm/README.md).

**Validation scope.** Both figures were rendered and visually checked after a fresh
full-grid baseline solve (all periods reported 100% convergence; zero invalid
simulated cells). The own-study specification passed finite-difference gradient,
time-budget, target-window and heterogeneous-simulation checks, the SMM self-test,
and a full-grid objective/fit evaluation. No new parameter estimation or complete
rerun of the notebook's later counterfactual sections was performed.

```julia
# -----------------------------
# Extraction function to compute average time series by group
# -----------------------------
# NaN-ROBUST. Plain `mean` over a slice containing a single NaN returns NaN, and a NaN
# series plots as nothing at all -- which is why baseline lines could silently vanish
# from the counterfactual panels. The child arrays legitimately carry NaN (agents on the
# other path, periods before a branch opens), so the group mean has to skip them.
nanmean(v) = (w = filter(isfinite, v); isempty(w) ? NaN : mean(w))

function extract_simulation_by_path(model, path_choice; warn_negative_assets::Bool = true)
    T = model.T
    college_idx = findall(x -> x == :college, path_choice)
    work_idx    = findall(x -> x == :work,    path_choice)

    if isempty(college_idx) || isempty(work_idx)
        @warn "extract_simulation_by_path: a path group is empty" n_college=length(college_idx) n_work=length(work_idx)
    end

    grp(arr, idx) = [nanmean(@view arr[idx, t]) for t in 1:T]

    out = (
      c      = (college = grp(model.sim_c, college_idx),      work = grp(model.sim_c, work_idx)),
      a      = (college = grp(model.sim_a, college_idx),      work = grp(model.sim_a, work_idx)),
      h      = (college = grp(model.sim_h, college_idx),      work = grp(model.sim_h, work_idx)),
      k      = (college = grp(model.sim_k, college_idx),      work = grp(model.sim_k, work_idx)),
      income = (college = grp(model.sim_income, college_idx), work = grp(model.sim_income, work_idx)),
      wage   = (college = grp(model.sim_wage, college_idx),   work = grp(model.sim_wage, work_idx))
    )

    # Assets below a_min are NOT a plotting artifact: snap_parent only corrects
    # float-sized violations (tol 1e-10), so a genuinely negative asset is a genuinely
    # negative simulated value -- an agent whose transfer did not cover tuition plus
    # consumption. Surface it rather than letting it show up silently as a plotted line.
    # Threshold is 1e-3, not float epsilon. Assets run 0-150 here, so a violation of
    # ~1e-4 is interpolation slack where the constraint binds, not an economic negative --
    # warning on those just trains you to ignore the warning. Anything the eye could see
    # on a plot is orders of magnitude above this.
    if warn_negative_assets
        fin  = filter(isfinite, vec(model.sim_a))
        worst = isempty(fin) ? 0.0 : minimum(fin)
        if worst < model.a_min - 1e-3
            neg = count(<(model.a_min - 1e-3), fin)
            @warn "Simulated child assets are genuinely below a_min" count=neg share=round(100neg/length(fin), digits=3) worst=round(worst, digits=4) a_min=model.a_min
        end
    end
    return out
end
```

```julia
baseline_sim = extract_simulation_by_path(child_model, path_choice_child)

function plot_simulation_results(
    model, path_choice, common_title="Counterfactual (r increased from X% to Y%)";
    baseline_sim=nothing,
    save::Bool = true,
    save_dir::AbstractString = nothing,   # pass a folder or it’ll auto-timestamp
    save_without_title::Bool = true
)
    # --- choose folder ---
    if save_dir === nothing
        stamp = Dates.format(Dates.now(), "yyyy-mm-dd_HHMMSS")
        save_dir = figdir("Adulthood", stamp)
    end

    # --- data & helpers ---
    sim_results = extract_simulation_by_path(model, path_choice)
    t = 1:model.T
    college_color, work_color = :blue, :red
    # BASELINE VISIBILITY. Baseline used the SAME colour as the counterfactual, dashed,
    # THINNER (1.0 vs 1.5) and drawn AFTER it -- so wherever the two nearly coincide the
    # dashed line sat on top of an identical-coloured solid line and read as one line.
    # Now: baseline is drawn FIRST (underneath), thicker and semi-transparent, so it
    # shows as a faint band behind the solid counterfactual even where they overlap.
    main_lw, base_lw, base_alpha = 2.0, 4.0, 0.30
    thousand_formatter = (val -> @sprintf("%.0f", val * 10))
    style_grid!(p) = plot!(p, grid=:on, gridstyle=:dash, gridcolor=:gray, gridalpha=0.5)

    ## ------------------- PLOT 1: Consumption ------------------- ##
    # include the BASELINE in the y-range, or a baseline above the counterfactual is clipped away
    ymax_c = maximum(filter(isfinite, vcat(sim_results.c.college, sim_results.c.work,
             baseline_sim === nothing ? Float64[] :
             vcat(baseline_sim.c.college, baseline_sim.c.work)))) * 1.05
    p1 = plot(t, sim_results.c.college,
        label = "College", lw = main_lw, color = college_color,
        xlabel = "Time", ylabel = "Consumption (×10³)", ylims = (0, ymax_c),
        title  = "Consumption", legend = :bottomright,
    )
    plot!(p1, t, sim_results.c.work, label = "Work", lw = main_lw, color = work_color)
    if baseline_sim !== nothing
        plot!(p1, t, baseline_sim.c.college, label = "College (Base)", lw = base_lw,
              color = college_color, linestyle = :dash, alpha = base_alpha)
        plot!(p1, t, baseline_sim.c.work,    label = "Work (Base)",    lw = base_lw,
              color = work_color,    linestyle = :dash, alpha = base_alpha)
    end
    plot!(p1, yformatter = thousand_formatter)
    style_grid!(p1)

    ## ------------------- PLOT 2: Assets ------------------- ##
    # Assets can be genuinely NEGATIVE (see extract_simulation_by_path): snap_parent only
    # fixes float-sized violations, so a transfer that fails to cover tuition shows up as
    # a real negative. Let the axis include 0 and any negative region rather than
    # implying the series is bounded below at zero.
    a_all = filter(isfinite, vcat(sim_results.a.college, sim_results.a.work,
                  baseline_sim === nothing ? Float64[] :
                  vcat(baseline_sim.a.college, baseline_sim.a.work)))
    a_lo = isempty(a_all) ? 0.0 : min(0.0, minimum(a_all) * 1.05)
    a_hi = isempty(a_all) ? 1.0 : maximum(a_all) * 1.05
    p2 = plot(t, sim_results.a.college, label = "College", lw = main_lw, color = college_color,
        xlabel = "Time", ylabel = "Assets (×10³)", title = "Assets", ylims = (a_lo, a_hi))
    a_lo < 0 && hline!(p2, [0.0], label = "", color = :black, lw = 0.8, alpha = 0.6)
    plot!(p2, t, sim_results.a.work, label = "Work", lw = main_lw, color = work_color)
    if baseline_sim !== nothing
        plot!(p2, t, baseline_sim.a.college, label = "College (Base)", lw = base_lw,
              color = college_color, linestyle = :dash, alpha = base_alpha)
        plot!(p2, t, baseline_sim.a.work,    label = "Work (Base)",    lw = base_lw,
              color = work_color,    linestyle = :dash, alpha = base_alpha)
    end
    plot!(p2, yformatter = thousand_formatter)
    style_grid!(p2)

    ## ------------------- PLOT 3: Hours ------------------- ##
    p3 = plot(t, sim_results.h.college, label = "College", lw = main_lw, color = college_color,
        xlabel = "Time", ylabel = "Hours", ylim = (0, 0.5), title = "Hours")   # hours are a share, (0,0.5) is a real bound
    plot!(p3, t, sim_results.h.work, label = "Work", lw = main_lw, color = work_color)
    if baseline_sim !== nothing
        plot!(p3, t, baseline_sim.h.college, label = "College (Base)", lw = base_lw,
              color = college_color, linestyle = :dash, alpha = base_alpha)
        plot!(p3, t, baseline_sim.h.work,    label = "Work (Base)",    lw = base_lw,
              color = work_color,    linestyle = :dash, alpha = base_alpha)
    end
    style_grid!(p3)

    ## ------------------- PLOT 4: Human Capital ------------------- ##
    p4 = plot(t, sim_results.k.college, label = "College", lw = main_lw, color = college_color,
        xlabel = "Time", ylabel = "Human Capital", title = "Human Capital")   # ylims (0,30) removed: it clipped the series out of view
    plot!(p4, t, sim_results.k.work, label = "Work", lw = main_lw, color = work_color)
    if baseline_sim !== nothing
        plot!(p4, t, baseline_sim.k.college, label = "College (Base)", lw = base_lw,
              color = college_color, linestyle = :dash, alpha = base_alpha)
        plot!(p4, t, baseline_sim.k.work,    label = "Work (Base)",    lw = base_lw,
              color = work_color,    linestyle = :dash, alpha = base_alpha)
    end
    style_grid!(p4)

    ## ------------------- PLOT 5: Income ------------------- ##
    p5 = plot(t, sim_results.income.college, label = "College", lw = main_lw, color = college_color,
        xlabel = "Time", ylabel = "Income (×10³)", title = "Income")
    plot!(p5, t, sim_results.income.work, label = "Work", lw = main_lw, color = work_color)
    if baseline_sim !== nothing
        plot!(p5, t, baseline_sim.income.college, label = "College (Base)", lw = base_lw,
              color = college_color, linestyle = :dash, alpha = base_alpha)
        plot!(p5, t, baseline_sim.income.work,    label = "Work (Base)",    lw = base_lw,
              color = work_color,    linestyle = :dash, alpha = base_alpha)
    end
    plot!(p5, yformatter = thousand_formatter)
    style_grid!(p5)

    ## ------------------- PLOT 6: Wage ------------------- ##
    p6 = plot(t, sim_results.wage.college, label = "College", lw = main_lw, color = college_color,
        xlabel = "Time", ylabel = "Wage", title = "Wage")
    plot!(p6, t, sim_results.wage.work, label = "Work", lw = main_lw, color = work_color)
    if baseline_sim !== nothing
        plot!(p6, t, baseline_sim.wage.college, label = "College (Base)", lw = base_lw,
              color = college_color, linestyle = :dash, alpha = base_alpha)
        plot!(p6, t, baseline_sim.wage.work,    label = "Work (Base)",    lw = base_lw,
              color = work_color,    linestyle = :dash, alpha = base_alpha)
    end
    style_grid!(p6)

    # --- combined plot for display (with title) ---
    combined_plot_display = plot(
        p1, p2, p3, p4, p5, p6;
        layout = (3, 2), size = (1000, 1300),
        plot_title = common_title, titlefont = 16, margin = 5mm, dpi = 100
    )

    # --- combined plot for saving (no title) ---
    combined_plot_save = save_without_title ?
        plot(p1, p2, p3, p4, p5, p6; layout = (3, 2), size = (1000, 1300), margin = 5mm, dpi = 100) :
        combined_plot_display

    # --- save with filename derived from the title ---
    if save
        isdir(save_dir) || mkpath(save_dir)
        safe_title = replace(replace(common_title, r"[^\w\s-]" => ""), " " => "_")
        filepath = joinpath(save_dir, safe_title * ".pdf")  # PDF only

        # if a file with same name exists, append a counter
        if isfile(filepath)
            i = 2
            while isfile(joinpath(save_dir, "$(safe_title)_$(i).pdf"))
                i += 1
            end
            filepath = joinpath(save_dir, "$(safe_title)_$(i).pdf")
        end

        savefig(combined_plot_save, filepath)  # vector PDF export
        @info "Saved plot as" filepath
    else
        @info "Plot not saved (save=false)"
    end

    return combined_plot_display
end


plot_simulation_results(
    child_model,
    path_choice_child,
    "Baseline Simulation: College vs. Work Paths";
    save = false,
    save_dir = figdir("Baseline"),
    save_without_title = true
)
```

```julia
# --- Extraction helper (unchanged) ---
function extract_all(model)
    return (
        c      = mean(model.sim_c, dims=1)[:],
        a      = mean(model.sim_a, dims=1)[:],      # length T+1
        h      = mean(model.sim_h, dims=1)[:],
        k      = mean(model.sim_k, dims=1)[:],
        t      = mean(model.sim_t, dims=1)[:],
        e      = mean(model.sim_e, dims=1)[:],
        i      = mean(model.sim_i, dims=1)[:],
        hc     = mean(model.sim_hc, dims=1)[:],     # length T+1
        income = mean(model.sim_income, dims=1)[:],
        wage   = mean(model.sim_wage, dims=1)[:]
    )
end

# --- Grid styling helper (unchanged) ---
function style_grid!()
    plot!(grid = :on, gridstyle = :dash, gridcolor = :gray, gridalpha = 0.4, minorgrid = false)
end


function plot_family_counterfactuals(
    models::Vector,
    labels::Vector{<:AbstractString};
    plot_title = nothing,
    colors = nothing,
    linestyles = nothing,
    lw_main = 1.5,
    lw_sub = 1.0,
    xtick_space = 2,
    save::Bool = true,
    save_dir::Union{Nothing,AbstractString} = nothing,
    dpi::Int = 100,
    fig_size::Tuple{Int,Int} = (1200, 1200)
)
    @assert length(models) == length(labels) "models and labels must have the same length."

    style_grid!() = plot!(
        grid=:on, gridstyle=:dash,
        foreground_color_grid=RGBA(0.5,0.5,0.5,0.85),
        gridlinewidth=1, minorgrid=false
    )

    function sanitize_filename(x)
        s = string(x)
        s = replace(s, r"\$" => "", "\\" => " ", "{" => "", "}" => "", "^" => "", ":" => "-", "/" => "-")
        s = replace(s, r"\s+" => "_")
        s = replace(s, r"[^\w\-\_]" => "")
        s = replace(s, r"_{2,}" => "_"); s = replace(s, r"-{2,}" => "-")
        isempty(s) ? "plot" : s
    end

    if save_dir === nothing
        stamp = Dates.format(Dates.now(), "yyyy-mm-dd_HHMMSS")
        save_dir = figdir("Parameters", stamp)
    end
    # NOTE: save_dir is only created inside `if save` below, so calling this
    # function with save=false no longer leaves an empty timestamped folder.
    n = length(models)
    extracts = [extract_all(m) for m in models]
    T = models[1].T
    t = 1:T
    t_full = 1:(T+1)

    if isnothing(colors)
        default_colors = [:black, :blue, :red, :green, :purple, :orange, :magenta, :teal]
        colors = [default_colors[mod1(i, length(default_colors))] for i in 1:n]
    end
    if isnothing(linestyles)
        linestyles = [:solid; repeat([:dash], n-1)...]
    end

    thousand_formatter2 = y -> string(round(y; digits=2))
    get_series(e, varsym) = varsym === :exp ? (e[:c] .+ e[:e]) : e[varsym]
    get_max_any(varsym)   = maximum(vcat([get_series(e, varsym) for e in extracts]...))
    get_max_full(varsym)  = maximum(vcat([e[varsym] for e in extracts]...)[1:end-1])

    ymax_c       = get_max_any(:c)      * 11
    ymax_a       = get_max_any(:a)      * 11
    ymax_income  = get_max_any(:income) * 11
    ymax_wage    = get_max_any(:wage)   * 1.05
    ymax_e       = get_max_any(:e)      * 11
    ymax_hc      = get_max_full(:hc)    * 1.1
    ymax_k       = get_max_any(:k)      * 1.1
    ymax_h       = get_max_any(:h)      * 1.1
    ymax_exp     = get_max_any(:exp)    * 11

    function scaled_series(e, varsym; full::Bool=false)
        s = get_series(e, varsym)
        scale10 = varsym in (:c, :a, :e, :income, :exp) ? 10.0 : 1.0
        return full ? s .* scale10 : s[1:T] .* scale10
    end

    function plot_variable(varsym; ylabel="", yformatter=nothing, ylim=nothing, marker=false, full=false)
        p = plot(dpi=dpi, size=fig_size)
        for i in 1:n
            lw   = i == 1 ? lw_main : lw_sub
            α    = i == 1 ? 1.0     : 0.9
            ms   = (marker && i == 1) ? :circle : :none
            xval = full ? t_full : t
            yval = scaled_series(extracts[i], varsym; full=full)
            plot!(p, xval, yval; label=labels[i], color=colors[i], lw=lw,
                  linestyle=linestyles[i], alpha=α, marker=ms)
        end
        plot!(p, xlabel="Time", ylabel=ylabel, ylim=ylim,
              xticks=1:xtick_space:(full ? T+1 : T), minorgrid=false)
        if !isnothing(yformatter); plot!(p, yformatter=yformatter); end
        style_grid!()
        return p
    end

    p1  = plot_variable(:c,   ylabel="Consumption (×10³)",      yformatter=thousand_formatter2, ylim=(0, ymax_c))
    p2  = plot_variable(:a,   ylabel="Assets (×10³)",           yformatter=thousand_formatter2, ylim=(0, ymax_a), full=true)
    p3  = plot_variable(:h,   ylabel="Hours Worked",            ylim=(0, ymax_h))
    p4  = plot_variable(:k,   ylabel="Parental HC",             ylim=(0, ymax_k))
    p5  = plot_variable(:e,   ylabel="Education Exp. (×10³)",   yformatter=thousand_formatter2, ylim=(0, ymax_e))
    p6  = plot_variable(:hc,  ylabel="Child's HC",              ylim=(0, ymax_hc), full=true)
    p7  = plot_variable(:t,   ylabel="Childcare Time",          ylim=(0, 0.7))
    p8  = plot_variable(:income, ylabel="Income (×10³)",        yformatter=thousand_formatter2, ylim=(0, ymax_income))
    p9  = plot_variable(:wage, ylabel="Wage",                   ylim=(0.0, max(0.1, ymax_wage)))
    p10 = plot_variable(:i,   ylabel="Child's Study Time",            yformatter=thousand_formatter2, ylim=(0, 0.7))
    p11 = plot_variable(:exp, ylabel="Total Exp. (×10³)",        yformatter=thousand_formatter2, ylim=(0, ymax_exp))

    display_title = isnothing(plot_title) ? "Simulation: " * join(labels, " vs ") : plot_title

    combined_plot_display = plot(
        p1, p2, p3, p5, p9, p7, p8, p10, p6;
        layout=(4, 3),
        size=fig_size,
        plot_title=display_title,
        titlefont=16,
        margin=5mm, dpi=dpi
    )

    # saved version (no title)
    combined_plot_save = plot(
        p1, p2, p3, p5, p9, p7, p8, p10, p6;
        layout=(3, 3),
        size=fig_size,
        margin=5mm, dpi=dpi
    )

    display(combined_plot_display)

    if save
        isdir(save_dir) || mkpath(save_dir)   # create only when actually saving
        name_source = isnothing(plot_title) ? display_title : plot_title
        safe_title = sanitize_filename(name_source)
        filepath = joinpath(save_dir, safe_title * ".pdf")  # ← PDF only

        if isfile(filepath)
            i = 2
            while isfile(joinpath(save_dir, "$(safe_title)_$(i).pdf")); i += 1; end
            filepath = joinpath(save_dir, "$(safe_title)_$(i).pdf")
        end

        savefig(combined_plot_save, filepath)  # vector PDF
        @info "Saved plot as" filepath
    else
        @info "Plot not saved (save=false)"
    end

    return combined_plot_display
end
```

```julia
# Function to compute average welfare at t=1 over initial simulated agents
function compute_average_welfare(model::Parent_child_interaction_age_specific_AR1)
    # Build interpolators for each shock state at t=1
    interp_dict = Dict{Int, Any}()
    for i_p in 1:model.Np
        vview = @view model.sol_v[1, :, :, :, i_p]  # Slice for t=1, this shock
        itp = interpolate((model.a_grid, model.k_grid, model.hc_grid), vview, Gridded(Linear()))
        interp_dict[i_p] = extrapolate(itp, Flat())
    end
    
    # Interpolate welfare for each simulated agent
    welfare_values = zeros(model.simN)
    for i in 1:model.simN
        a_init = model.sim_a_init[i]
        k_init = model.sim_k_init[i]
        hc_init = model.sim_hc_init[i]
        p_idx = model.sim_p_init[i]
        
        # Clamp to grid bounds to avoid extrapolation issues
        a_clamp = clamp(a_init, model.a_min, model.a_max)
        k_clamp = clamp(k_init, model.k_min, model.k_max)
        hc_clamp = clamp(hc_init, model.hc_min, model.hc_max)
        
        welfare_values[i] = interp_dict[p_idx](a_clamp, k_clamp, hc_clamp)
    end
    
    return mean(welfare_values)
end

# Compute average welfare for each policy
avg_welfare_baseline = compute_average_welfare(model_baseline)
avg_welfare_tax_high = compute_average_welfare(model_tax_benefit_high)
avg_welfare_tax_low = compute_average_welfare(model_tax_benefit_low)

# N9: the printed tax rates were stale, carried over from transfer_model_AR1.ipynb where
# the baseline really was 0.25 and the high arm 0.35. Here the baseline is the constructor
# default and the arms are whatever cell 28 passed, so they are read off the models.
tau_base = model_baseline.tau
tau_hi   = model_tax_benefit_high.tau
tau_lo   = model_tax_benefit_low.tau
y_base, y_hi, y_lo = model_baseline.y, model_tax_benefit_high.y, model_tax_benefit_low.y

@printf("Average Welfare (Baseline,  tau=%.2f, b=%.0f): %.6f\n", tau_base, 1000*y_base, avg_welfare_baseline)
@printf("Average Welfare (High tax,  tau=%.2f, b=%.0f): %.6f\n", tau_hi,   1000*y_hi,   avg_welfare_tax_high)
@printf("Average Welfare (Low tax,   tau=%.2f, b=%.0f): %.6f\n", tau_lo,   1000*y_lo,   avg_welfare_tax_low)

# ---------------------------------------------------------------------------
# N3: the consumption-equivalent variation has been REMOVED, not repaired.
#
# Both the old forms assumed the value function is homogeneous of degree (1-rho) in
# consumption:
#     lambda = exp(dV / (phi1 * sum beta^t)) - 1          (log-utility version)
#     lambda = (V_cf / V_base)^(1/(1-rho)) - 1            (this notebook's version)
# V here is NOT homothetic in consumption: it contains alpha_2*log(HC),
# alpha_1*log(l_c), -phi_2*h^(1+eta)/(1+eta) and a terminal value
# psi*log(HC) + kappa*log(a_term) + omega*V^C. Scaling consumption does not scale any of
# those, so no exponent recovers a consumption equivalent and the number has no
# interpretation. See N3 in docs/ERRORS.md.
#
# What IS reported instead: the raw welfare difference, which is well defined, plus a
# bootstrap standard error so the comparison can be read against Monte Carlo noise.
# A valid CEV would require solving for the lambda that equates
# V(baseline with consumption scaled by 1+lambda) to V(counterfactual) -- a root-find over
# a re-solve, not a closed form. Left undone deliberately.
# ---------------------------------------------------------------------------
welfare_gap_high = avg_welfare_tax_high - avg_welfare_baseline
welfare_gap_low  = avg_welfare_tax_low  - avg_welfare_baseline
@printf("\nWelfare difference vs baseline (utils, NOT a consumption equivalent):\n")
@printf("  High tax & benefit: %+.6f\n", welfare_gap_high)
@printf("  Low  tax & benefit: %+.6f\n", welfare_gap_low)
```

```julia
function plot_terminal_asset_comparison(
    models::Vector,
    labels::Vector{<:AbstractString};
    bins::Int = 40,
    rescale::Real = 10,
    terminal_idx::Int = 18,
    plot_title = "Terminal Asset Distribution",
    save::Bool = true,
    save_dir::AbstractString = figdir("Parameters", "terminal_assets")  # Default folder
)
    # --- helpers ---
    sanitize_filename(x) = begin
        s = string(x)
        s = replace(s, r"\$" => "", "\\" => " ", "{" => "", "}" => "", "^" => "", ":" => "-", "/" => "-")
        s = replace(s, r"\s+" => "_")
        s = replace(s, r"[^\w\-\_]" => "")
        s = replace(s, r"_{2,}" => "_"); s = replace(s, r"-{2,}" => "-")
        isempty(s) ? "plot" : s
    end

    # Extract asset data for each model
    assets_list = [model.sim_a[:, terminal_idx] .* rescale for model in models]
    plot_labels = reshape(labels, 1, :)

    # --- individual panels ---
    p1 = histogram(
        assets_list;
        bins = bins,
        xlabel = "Parental Asset (×10³)",
        ylabel = "Count",
        legend = :topright,
        label = plot_labels,
        opacity = 0.5,
        left_margin = 15mm,
        bottom_margin = 12mm
    )

    p2 = density(
        assets_list;
        xlabel = "Parental Asset (×10³)",
        ylabel = "Density",
        legend = :topright,
        label = plot_labels,
        lw = 2,
        left_margin = 2mm,
        bottom_margin = 12mm
    )

    # --- combined plot: display version (with title) ---
    plt_display = plot(
        p1, p2;
        layout = (1, 2),
        size = (1100, 450),
        plot_title = plot_title,
        plot_titlefontsize = 14,
        top_margin = 4mm,
        guidefont = font(12),
        tickfont = font(10),
        legendfont = font(7),
        dpi = 100
    )
    display(plt_display)

    # --- save version (no title) ---
    if save
        plt_save = plot(
            p1, p2;
            layout = (1, 2),
            size = (1100, 450),
            top_margin = 4mm,
            guidefont = font(12),
            tickfont = font(10),
            legendfont = font(7),
            dpi = 100
        )

        # ---- Save as PDF ----
        isdir(save_dir) || mkpath(save_dir)
        safe_title = sanitize_filename(plot_title)
        outpath = joinpath(save_dir, safe_title * ".pdf")
        if isfile(outpath)
            i = 2
            while isfile(joinpath(save_dir, "$(safe_title)_$(i).pdf")); i += 1; end
            outpath = joinpath(save_dir, "$(safe_title)_$(i).pdf")
        end
        savefig(plt_save, outpath)  # vector PDF
        @info "Saved plot to" outpath
    end

    return plt_display
end
```

## Subjective Expectations

---
  **Terminal Value at Period $T_L$ and College Decision Half period**

  The parent utility in period ($T = T_L$) for an family with belief $m$ is defined as: 
  $$
  V_{T_L}^P = \underbrace{\psi_{\text{term}} \cdot \log HC_{T_L}}_{\text{HC Prestige}} + \kappa_{\text{term}} \cdot \log a_{term} + \underbrace{\omega \cdot V_{T_L}^C}_{\text{Altruism}}
  $$

The child is discrete choice is $d \in \{E,W\}$:

  $$
  V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L}, m) = \max \left\{ V_0^E(tr, HC_{T_L}, \epsilon_0, z_{T_L}, m), V_0^W(tr, HC_{T_L}, z_{T_L}) \right\}
  $$



The family’s ex-ante value function 


$$
V_{T_L}^{\mathrm{CD}}(a_{T_L}, HC_{T_L}, \epsilon_0, m) =
\max_{0 \leq tr \leq a_{T_L}} \left\{
\mathbb{E}_{z_{T_L}} \left[
(1 - \theta) V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L}, m)
+ \theta V_{T_L}^P(tr, HC_{T_L}, \epsilon_0, m)
\right]
\right\}
$$



- in period $T_L-1$ we had:
    $$
    V_{T_L-1}^{\mathrm{CD}}(a_{T_L}, HC_{T_L}, m) =
    \max_{0 \leq tr\leq } \left\{
        \mathbb{E}_{\epsilon_0,\, z_{T_L}} \left[
            (1 - \theta) V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L}, m)
            + \theta V_{T_L}^P(tr, HC_{T_L}, \epsilon_0, m)
        \right]
    \right\}
    $$

The decision will be: 
$$
d^*(a_{T_L}, HC_{T_L}, \epsilon_0) =
     \arg\max_{0 \leq tr \leq a_{T_L}}\left\{
            \mathbb{E}_{z_{T_L}} \left[
                (1-\theta)\, V_{T_L}^C(tr, HC_{T_L}, \epsilon_0, z_{T_L}, m)
                + \theta\, V_{T_L}^P(tr, a_{T_L}, HC_{T_L}, \epsilon_0, m)            \right]
        \right\}
$$

subject to the following constraints:



- **Budget constraint**:
  $$
  a_{term} = a_{T_L} - tr
  $$

$$

$$
a_{t+1} = (1 + r) a_t - c_t - c_{\text{college}} + b_t

\end{align}

## single biased belief

## Beliefs: from the Bleemer–Zafar survey number to the model's `beta_E`

**What the survey gives us.** Bleemer and Zafar (2018) elicit beliefs about the *earnings
return* to college: respondents report how much more a college graduate earns, as a single
ratio. Call it `rce` (relative college earnings) — `rce = 1.5` means "a graduate earns 50%
more". In this notebook the belief distribution is `rce ∈ [1, 3]`, built as `1 + 2z` with
`z ~ Beta(2, 5)` (right-skewed: most people believe a modest premium, a tail believes a
large one).

**What the model needs.** The wage equation does not carry a single premium. It carries a
*profile* in age, from Daruich and Fernández (2023):

$$\ln w = \ln w_0 + \underbrace{\beta_E + \gamma_{1E}\,\text{age} + \gamma_{2E}\,\text{age}^2}_{\text{college premium at that age}} + (\alpha_\theta + \alpha_{\theta E})\ln\theta + \dots$$

The estimated premium runs from **0.25 at age 22 to 0.51 at age 50**, so a scalar belief
cannot be read off any single age — it has to be anchored to a horizon.

**The transformation.** We anchor on the **unweighted career average** over the graduate's
working life, ages `t_college+1 … T`, which is the natural reading of "relative college
earnings" as one number covering a career. A believer's profile is the estimated one shifted
in parallel:

$$\beta_E^m = \beta_E^\star + \big[\log(\text{rce}_m) - \text{anchor}^\star\big],
\qquad \text{anchor}^\star = \beta_E^\star + \overline{\gamma_{1E}a + \gamma_{2E}a^2}$$

The estimated intercept $\beta_E^\star$ **cancels**, leaving

$$\boxed{\;\beta_E^m \;=\; \log(\text{rce}_m)\;-\;\overline{\gamma_{1E}\,a + \gamma_{2E}\,a^2}\;}$$

implemented as `beta_E_from_rce(model, rce)` in `child_lifecycle.jl`. Two properties matter:

1. The mapping depends **only on the interaction terms**, not on where the estimated
   intercept happens to sit — so re-estimating the wage equation does not silently move the
   belief scale.
2. Beliefs shift the **level** of the premium and leave its **shape** at the estimated one.
   A believer is someone who is wrong about how big the return is, not about how it evolves
   with age.

**Why not the old mapping.** This replaces `college_boost = (rce - 1)/0.4`, which inverted
the previous `w0·(1 + α·HC)` wage at `HC = 0` to re-express the belief as a *human-capital
increment*. Under a log wage that inversion is unnecessary — the belief is already in the
model's units — and it is also further from what the survey actually asks, which is about
earnings, not about human capital.

**Worked example.** With `rce = 1.9` ("a graduate earns 90% more"): `log(1.9) = 0.642`, minus
the career-average interaction term, gives the `beta_E` used by that belief type.

```julia
# ---------------------------
# 2. Subjective Model (Single Belief Value but Keep Original Structure)
# ---------------------------

ProgressMeter.ijulia_behavior(:clear)

# ========== STEP 0: Parameters ==========
simN = 5000

# ---- Single belief value (everyone shares this) ----
single_rce_belief = 1.9   # <--- believed college/high-school earnings ratio

# One-element vector, preserving the original indexing
num_bins = 1

# Everyone belongs to bin 1
belief_type_s = fill(1, simN)

# ========== STEP 2: Solve Work Problem Once ==========
base_child = ConSavLaborCollege_AR1(; Na=30, Nk=30, Nt=5, CHILD_DEFAULTS...
)
@suppress_output solve_model_work!(base_child)
optimal_transfer_work!(base_child)
# base_child carries the TRUE beta_E. Its COLLEGE branch is needed too, because a
# graduate's working life is simulated with the true policy: the belief governs the
# college-vs-work decision and the college years, but on graduating the agent observes
# the real wage and re-optimizes against it. simulate_model_family_hetero! reads
# base_child.sol_c_grad / sol_h_grad for those years and errors if they are missing.
@suppress_output solve_model_college!(base_child)
optimal_transfer_college!(base_child)

# See the multi-bin cells: the belief is the log college wage premium.
beta_E_belief_bin = [beta_E_from_rce(base_child, single_rce_belief)]

# ========== STEP 3: Belief-Specific Child Models ==========
child_models = Vector{ConSavLaborCollege_AR1}(undef, num_bins)
V_child_interps = Vector{Any}(undef, num_bins)

@showprogress dt=0.1 "Creating and solving child models..." for m in 1:num_bins
    # SOFT SCOPE. This was `child_model = ...`, which SILENTLY CLOBBERED the global
    # `child_model` holding the simulated baseline (cell 21): a top-level `for` in Julia
    # uses soft scope, so assigning a name that already exists globally rebinds the
    # global. By the counterfactual plots, `child_model` was the LAST belief bin's model
    # -- solved but never simulated, so its sim arrays were still all NaN and
    # `baseline_sim` came out NaN. An all-NaN series still produces a legend entry but
    # draws nothing, which is exactly why the "(Base)" lines were missing from every
    # counterfactual panel. Renamed to a loop-local name.
    belief_child = ConSavLaborCollege_AR1(; Na=30, Nk=30, Nt=5, CHILD_DEFAULTS...,
        beta_E=beta_E_belief_bin[m]
    )
    # Copy work solution
    belief_child.sol_c_work     = base_child.sol_c_work
    belief_child.sol_h_work     = base_child.sol_h_work
    belief_child.sol_v_work     = base_child.sol_v_work
    belief_child.sol_tr_v_work  = base_child.sol_tr_v_work
    belief_child.sol_tr_work    = base_child.sol_tr_work

    # Solve college-specific parts
    @suppress_output solve_model_college!(belief_child)
    optimal_transfer_college!(belief_child)
    # N1 (Phase 0.5 timing): eps_0 is observed BEFORE the transfer, so the terminal
    # object is E_eps[ max_d ( max_tr E_z[ W_d ] ) ] -- the discrete college/work max
    # sits INSIDE the expectation over eps, not outside it. terminal_value_spline
    # assembles exactly that, fitted over the rows where the value is finite.
    V_child_interp = terminal_value_spline(belief_child; s=10.0)

    child_models[m] = belief_child
    V_child_interps[m] = V_child_interp
end

# ========== STEP 4: Create and Solve Parent Models ==========
parent_models_s = Vector{Parent_child_interaction_age_specific_AR1}(undef, num_bins)
@showprogress dt=0.1 "Creating and solving parent models..." for m in 1:num_bins
    parent_models_s[m] = Parent_child_interaction_age_specific_AR1(
        Na=30, Nk=2, Nhc=30, # Add more params as needed
    )
    parent_models_s[m].V_child_interp = V_child_interps[m]
    @suppress_output solve_model!(parent_models_s[m])
end
```

```julia
# ---------------------------
# 2. Subjective Model
# ---------------------------

ProgressMeter.ijulia_behavior(:clear)
# ========== STEP 0: Parameters ==========
simN = 5000
α, β = 2.0, 5.0
# N15: this draw was unseeded, so every counterfactual re-randomized the belief
# distribution and differences between arms carried resampling noise that P2 was
# supposed to remove. NB_SEED is defined once, in the setup cell.
z = rand(MersenneTwister(NB_SEED), Beta(α, β), simN)
rce_belief_vec = 1 .+ 2.0 .* z   # In [1, 3]

bin_edges = 1.0:0.1:3.0
num_bins = length(bin_edges) - 1

# Bin midpoints (for reporting/analysis, or to use for each bin's model)
rce_mid = [ (bin_edges[j] + bin_edges[j+1]) / 2 for j in 1:num_bins ]


# ========== STEP 1: Assign agents to belief bins ==========
belief_type = similar(rce_belief_vec, Int)
for i in 1:simN
    bin_idx = min(searchsortedfirst(bin_edges, rce_belief_vec[i]) - 1, num_bins)
    belief_type[i] = bin_idx < 1 ? 1 : bin_idx
end
# Now, for each agent, belief_type[i] is in 1:num_bins


# ========== STEP 2: Solve Work Problem Once ==========
base_child = ConSavLaborCollege_AR1(; Na=30, Nk=30, Nt=5, CHILD_DEFAULTS...
)
@suppress_output solve_model_work!(base_child)
optimal_transfer_work!(base_child)
# base_child carries the TRUE beta_E. Its COLLEGE branch is needed too, because a
# graduate's working life is simulated with the true policy: the belief governs the
# college-vs-work decision and the college years, but on graduating the agent observes
# the real wage and re-optimizes against it. simulate_model_family_hetero! reads
# base_child.sol_c_grad / sol_h_grad for those years and errors if they are missing.
@suppress_output solve_model_college!(base_child)
optimal_transfer_college!(base_child)

# Beliefs are about beta_E, the LOG COLLEGE WAGE PREMIUM. Bleemer and Zafar (2018)
# elicit the earnings return, so rce (the believed college/high-school earnings ratio)
# is already the right object; beta_E_from_rce anchors that single number to the career
# average of the premium profile, ages 22-68. Replaces (rce - 1)/0.4, which inverted the
# old w0*(1 + alpha*HC) wage at HC = 0 to re-express the belief as a stock increment.
beta_E_belief_bin = round.(beta_E_from_rce(base_child, rce_mid), digits=4)

# ========== STEP 3: Belief-Specific Child Models ==========
child_models = Vector{ConSavLaborCollege_AR1}(undef, num_bins)
V_child_interps = Vector{Any}(undef, num_bins)

@showprogress dt=0.1 "Creating and solving child models..." for m in 1:num_bins
    # SOFT SCOPE. This was `child_model = ...`, which SILENTLY CLOBBERED the global
    # `child_model` holding the simulated baseline (cell 21): a top-level `for` in Julia
    # uses soft scope, so assigning a name that already exists globally rebinds the
    # global. By the counterfactual plots, `child_model` was the LAST belief bin's model
    # -- solved but never simulated, so its sim arrays were still all NaN and
    # `baseline_sim` came out NaN. An all-NaN series still produces a legend entry but
    # draws nothing, which is exactly why the "(Base)" lines were missing from every
    # counterfactual panel. Renamed to a loop-local name.
    belief_child = ConSavLaborCollege_AR1(; Na=30, Nk=30, Nt=5, CHILD_DEFAULTS...,
        beta_E=beta_E_belief_bin[m]
    )
    # Copy work solution
    belief_child.sol_c_work     = base_child.sol_c_work
    belief_child.sol_h_work     = base_child.sol_h_work
    belief_child.sol_v_work     = base_child.sol_v_work
    belief_child.sol_tr_v_work  = base_child.sol_tr_v_work
    belief_child.sol_tr_work    = base_child.sol_tr_work

    # Solve college-specific parts
    @suppress_output solve_model_college!(belief_child)
    optimal_transfer_college!(belief_child)
    # N1 (Phase 0.5 timing): eps_0 is observed BEFORE the transfer, so the terminal
    # object is E_eps[ max_d ( max_tr E_z[ W_d ] ) ] -- the discrete college/work max
    # sits INSIDE the expectation over eps, not outside it. terminal_value_spline
    # assembles exactly that, fitted over the rows where the value is finite.
    V_child_interp = terminal_value_spline(belief_child; s=10.0)

    child_models[m] = belief_child
    V_child_interps[m] = V_child_interp
end

# ========== STEP 4: Create and Solve Parent Models ==========
parent_models = Vector{Parent_child_interaction_age_specific_AR1}(undef, num_bins)
@showprogress dt=0.1 "Creating and solving parent models..." for m in 1:num_bins
    parent_models[m] = Parent_child_interaction_age_specific_AR1(
        Na=30, Nk=2, Nhc=30, # Add more params as needed
    )
    parent_models[m].V_child_interp = V_child_interps[m]
    @suppress_output solve_model!(parent_models[m])
end

# Now, for simulation, use:
#   For agent i, their belief_type[i] gives the bin index,
#   and you can use parent_models[belief_type[i]], etc.
```

```julia
# -----------------------------
# Extraction function to compute average time series for whole population
# -----------------------------
function extract_simulation_population(model)
    T = model.T
    # Compute the average for each time period over all agents
    avg_c = [mean(model.sim_c[:, t]) for t in 1:T]
    avg_a = [mean(model.sim_a[:, t]) for t in 1:T]
    avg_h = [mean(model.sim_h[:, t]) for t in 1:T]
    avg_k = [mean(model.sim_k[:, t]) for t in 1:T]
    avg_income = [mean(model.sim_income[:, t]) for t in 1:T]
    avg_wage = [mean(model.sim_wage[:, t]) for t in 1:T]
    
    return (
      c      = avg_c,
      a      = avg_a,
      h      = avg_h,
      k      = avg_k,
      income = avg_income,
      wage   = avg_wage
    )
end

function plot_population_simulation_results(
    model, common_title="Counterfactual (r increased from X% to Y%)";
    baseline_sim=nothing,
    save::Bool = true,
    save_dir::AbstractString = nothing,   # pass a folder or it’ll auto-timestamp
    save_without_title::Bool = true
)
    # --- choose folder ---
    if save_dir === nothing
        stamp = Dates.format(Dates.now(), "yyyy-mm-dd_HHMMSS")
        save_dir = figdir("Adulthood", stamp)
    end

    # --- data & helpers ---
    sim_results = extract_simulation_population(model)
    t = 1:model.T
    current_color, base_color = :blue, :red
    main_lw, base_lw = 1.5, 1.0
    thousand_formatter = (val -> @sprintf("%.0f", val * 10))
    style_grid!(p) = plot!(p, grid=:on, gridstyle=:dash, gridcolor=:gray, gridalpha=0.5)

    ## ------------------- PLOT 1: Consumption ------------------- ##
    ymax_c = maximum(sim_results.c) * 1.05
    if baseline_sim !== nothing
        ymax_c = max(ymax_c, maximum(baseline_sim.c) * 1.05)
    end
    p1 = plot(t, sim_results.c,
        label = "Population", lw = main_lw, color = current_color,
        xlabel = "Time", ylabel = "Consumption (×10³)", ylims = (0, ymax_c),
        title  = "Consumption", legend = :bottomright,
    )
    if baseline_sim !== nothing
        plot!(p1, t, baseline_sim.c, label = "Population (Base)", lw = base_lw, color = base_color, linestyle = :dash)
    end
    plot!(p1, yformatter = thousand_formatter)
    style_grid!(p1)

    ## ------------------- PLOT 2: Assets ------------------- ##
    p2 = plot(t, sim_results.a, label = "Population", lw = main_lw, color = current_color,
        xlabel = "Time", ylabel = "Assets (×10³)", title = "Assets")
    if baseline_sim !== nothing
        plot!(p2, t, baseline_sim.a, label = "Population (Base)", lw = base_lw, color = base_color, linestyle = :dash)
    end
    plot!(p2, yformatter = thousand_formatter)
    style_grid!(p2)

    ## ------------------- PLOT 3: Hours ------------------- ##
    p3 = plot(t, sim_results.h, label = "Population", lw = main_lw, color = current_color,
        xlabel = "Time", ylabel = "Hours", ylim = (0, 0.5), title = "Hours")
    if baseline_sim !== nothing
        plot!(p3, t, baseline_sim.h, label = "Population (Base)", lw = base_lw, color = base_color, linestyle = :dash)
    end
    style_grid!(p3)

    ## ------------------- PLOT 4: Human Capital ------------------- ##
    p4 = plot(t, sim_results.k, label = "Population", lw = main_lw, color = current_color,
        xlabel = "Time", ylabel = "Human Capital", ylims = (0, 30), title = "Human Capital")
    if baseline_sim !== nothing
        plot!(p4, t, baseline_sim.k, label = "Population (Base)", lw = base_lw, color = base_color, linestyle = :dash)
    end
    style_grid!(p4)

    ## ------------------- PLOT 5: Income ------------------- ##
    p5 = plot(t, sim_results.income, label = "Population", lw = main_lw, color = current_color,
        xlabel = "Time", ylabel = "Income (×10³)", title = "Income")
    if baseline_sim !== nothing
        plot!(p5, t, baseline_sim.income, label = "Population (Base)", lw = base_lw, color = base_color, linestyle = :dash)
    end
    plot!(p5, yformatter = thousand_formatter)
    style_grid!(p5)

    ## ------------------- PLOT 6: Wage ------------------- ##
    p6 = plot(t, sim_results.wage, label = "Population", lw = main_lw, color = current_color,
        xlabel = "Time", ylabel = "Wage", title = "Wage")
    if baseline_sim !== nothing
        plot!(p6, t, baseline_sim.wage, label = "Population (Base)", lw = base_lw, color = base_color, linestyle = :dash)
    end
    style_grid!(p6)

    # --- combined plot for display (with title) ---
    combined_plot_display = plot(
        p1, p2, p3, p4, p5, p6;
        layout = (3, 2), size = (1000, 1300),
        plot_title = common_title, titlefont = 16, margin = 5mm, dpi = 100
    )

    # --- combined plot for saving (no title) ---
    combined_plot_save = save_without_title ?
        plot(p1, p2, p3, p4, p5, p6; layout = (3, 2), size = (1000, 1300), margin = 5mm, dpi = 100) :
        combined_plot_display

    # --- save with filename derived from the title ---
    if save
        isdir(save_dir) || mkpath(save_dir)
        safe_title = replace(replace(common_title, r"[^\w\s-]" => ""), " " => "_")
        filepath = joinpath(save_dir, safe_title * ".pdf")  # PDF only

        # if a file with same name exists, append a counter
        if isfile(filepath)
            i = 2
            while isfile(joinpath(save_dir, "$(safe_title)_$(i).pdf"))
                i += 1
            end
            filepath = joinpath(save_dir, "$(safe_title)_$(i).pdf")
        end

        savefig(combined_plot_save, filepath)  # vector PDF export
        @info "Saved plot as" filepath
    else
        @info "Plot not saved (save=false)"
    end

    return combined_plot_display
end

#
```

### Subjective with resource 

```julia
# ---------------------------
# 1. Expectation + Hetero Model
# ---------------------------

ProgressMeter.ijulia_behavior(:clear)
# ========== STEP 0: Parameters ==========
simN = 5000
α, β = 2.0, 5.0
# N15: this draw was unseeded, so every counterfactual re-randomized the belief
# distribution and differences between arms carried resampling noise that P2 was
# supposed to remove. NB_SEED is defined once, in the setup cell.
z = rand(MersenneTwister(NB_SEED), Beta(α, β), simN)
rce_belief_vec = 1 .+ 2.0 .* z   # In [1, 3]

bin_edges = 1.0:0.1:3.0
num_bins = length(bin_edges) - 1

# Bin midpoints (for reporting/analysis, or to use for each bin's model)
rce_mid = [ (bin_edges[j] + bin_edges[j+1]) / 2 for j in 1:num_bins ]


# ========== STEP 1: Assign agents to belief bins ==========
belief_type = similar(rce_belief_vec, Int)
for i in 1:simN
    bin_idx = min(searchsortedfirst(bin_edges, rce_belief_vec[i]) - 1, num_bins)
    belief_type[i] = bin_idx < 1 ? 1 : bin_idx
end
# Now, for each agent, belief_type[i] is in 1:num_bins


# ========== STEP 2: Solve Work Problem Once ==========
base_child = ConSavLaborCollege_AR1(; Na=30, Nk=30, Nt=5, CHILD_DEFAULTS...,
    y=1.8 * CHILD_DEFAULTS.y   # N10: base_child MUST carry the treatment y. Without it the
           # work path (and the post-college continuation, which copies the
           # work arrays) was solved at the default y=0.6 while college ran
           # at y=1.08, handing college a resource bonus work never got.
           # 2026-09-27: 1.08 = 1.8x the old 0.6; now 1.8x CHILD_DEFAULTS.y (0.144).
)
@suppress_output solve_model_work!(base_child)
optimal_transfer_work!(base_child)
# base_child carries the TRUE beta_E. Its COLLEGE branch is needed too, because a
# graduate's working life is simulated with the true policy: the belief governs the
# college-vs-work decision and the college years, but on graduating the agent observes
# the real wage and re-optimizes against it. simulate_model_family_hetero! reads
# base_child.sol_c_grad / sol_h_grad for those years and errors if they are missing.
@suppress_output solve_model_college!(base_child)
optimal_transfer_college!(base_child)

# Beliefs are about beta_E, the LOG COLLEGE WAGE PREMIUM. Bleemer and Zafar (2018)
# elicit the earnings return, so rce (the believed college/high-school earnings ratio)
# is already the right object; beta_E_from_rce anchors that single number to the career
# average of the premium profile, ages 22-68. Replaces (rce - 1)/0.4, which inverted the
# old w0*(1 + alpha*HC) wage at HC = 0 to re-express the belief as a stock increment.
beta_E_belief_bin = round.(beta_E_from_rce(base_child, rce_mid), digits=4)

# ========== STEP 3: Belief-Specific Child Models ==========
child_models_E = Vector{ConSavLaborCollege_AR1}(undef, num_bins)
V_child_interps = Vector{Any}(undef, num_bins)

@showprogress dt=0.1 "Creating and solving child models..." for m in 1:num_bins
    # SOFT SCOPE. This was `child_model = ...`, which SILENTLY CLOBBERED the global
    # `child_model` holding the simulated baseline (cell 21): a top-level `for` in Julia
    # uses soft scope, so assigning a name that already exists globally rebinds the
    # global. By the counterfactual plots, `child_model` was the LAST belief bin's model
    # -- solved but never simulated, so its sim arrays were still all NaN and
    # `baseline_sim` came out NaN. An all-NaN series still produces a legend entry but
    # draws nothing, which is exactly why the "(Base)" lines were missing from every
    # counterfactual panel. Renamed to a loop-local name.
    belief_child = ConSavLaborCollege_AR1(; Na=30, Nk=30, Nt=5, CHILD_DEFAULTS...,
        beta_E=beta_E_belief_bin[m], y=1.8 * CHILD_DEFAULTS.y
    )
    # Copy work solution
    belief_child.sol_c_work     = base_child.sol_c_work
    belief_child.sol_h_work     = base_child.sol_h_work
    belief_child.sol_v_work     = base_child.sol_v_work
    belief_child.sol_tr_v_work  = base_child.sol_tr_v_work
    belief_child.sol_tr_work    = base_child.sol_tr_work

    # Solve college-specific parts
    @suppress_output solve_model_college!(belief_child)
    optimal_transfer_college!(belief_child)
    # N1 (Phase 0.5 timing): eps_0 is observed BEFORE the transfer, so the terminal
    # object is E_eps[ max_d ( max_tr E_z[ W_d ] ) ] -- the discrete college/work max
    # sits INSIDE the expectation over eps, not outside it. terminal_value_spline
    # assembles exactly that, fitted over the rows where the value is finite.
    V_child_interp = terminal_value_spline(belief_child; s=10.0)

    child_models_E[m] = belief_child
    V_child_interps[m] = V_child_interp
end

# ========== STEP 4: Create and Solve Parent Models ==========
parent_models_E = Vector{Parent_child_interaction_age_specific_AR1}(undef, num_bins)
@showprogress dt=0.1 "Creating and solving parent models..." for m in 1:num_bins
    parent_models_E[m] = Parent_child_interaction_age_specific_AR1(
        Na=30, Nk=2, Nhc=30, # Add more params as needed
    )
    parent_models_E[m].V_child_interp = V_child_interps[m]
    @suppress_output solve_model!(parent_models_E[m])
end

# Now, for simulation, use:
#   For agent i, their belief_type[i] gives the bin index,
#   and you can use parent_models_E[belief_type[i]], etc.
```
